```python
import pandas as pd

# Load files
phishtank = pd.read_csv("../data/raw/phishtank.csv")
legit = pd.read_csv("../data/raw/legit.csv")
top1million = pd.read_csv("../data/raw/top1million.csv")

# Check column names
print("PhishTank columns:", phishtank.columns)
print("Legit columns:", legit.columns)
print("Top1Million columns:", top1million.columns)
```

```python
phishtank = phishtank.rename(columns={phishtank.columns[0]: "url"})
legit = legit.rename(columns={legit.columns[0]: "url"})
top1million = top1million.rename(columns={top1million.columns[0]: "url"})
```

```python
def normalize_url(url):
    url = str(url).strip().lower()
    
    if not url.startswith("http://") and not url.startswith("https://"):
        url = "https://" + url
    
    return url
```

```python
phishtank["url"] = phishtank["url"].apply(normalize_url)
legit["url"] = legit["url"].apply(normalize_url)
top1million["url"] = top1million["url"].apply(normalize_url)
```

```python
phishtank["label"] = 1
legit["label"] = 0
top1million["label"] = 0
```

```python
phishtank = phishtank[["url", "label"]].head(50000)
legit = legit[["url", "label"]].head(35000)
top1million = top1million[["url", "label"]].head(15000)
```

```python
df = pd.concat([phishtank, legit, top1million], ignore_index=True)
```

```python
df = df.drop_duplicates(subset="url")
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
print(df["label"].value_counts())
print(df.shape)
print(df.head())
```

```python
df.to_csv("../data/processed/final_phishing_dataset.csv", index=False, encoding="utf-8")
print("Dataset saved successfully.")
```
